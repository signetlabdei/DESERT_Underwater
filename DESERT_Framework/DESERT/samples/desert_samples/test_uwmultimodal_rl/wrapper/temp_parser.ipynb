```python
import os
import json
from matplotlib import pyplot as plt

PRR = []
Throughput = []
Jitter = []

results_folder = "res/deleteme/campaign_0"
if not os.path.isdir(results_folder):
    raise FileNotFoundError(f"Results folder not found: {results_folder}")

for entry in os.listdir(results_folder):
    filename = os.fsdecode(entry)
    if filename.endswith(".json"):
        path = os.path.join(results_folder, filename)
        try:
            with open(path, "r") as f:
                data = json.load(f)
        except json.JSONDecodeError:
            print("Warning: invalid JSON in", path)
            continue
        except Exception as e:
            print("Warning: error reading", path, "->", e)
            continue

        prr = data.get("Average PRR")
        if prr is None:
            print(f"Warning: 'Average PRR' missing in {path}")
        else:
            PRR.append(prr)

        thr = data.get("Average Instantaneous Throughput")
        if thr is None:
            print(f"Warning: 'Average Instantaneous Throughput' missing in {path}")
        else:
            Throughput.append(thr)

        jit = data.get("Average Jitter")
        if jit is None:
            print(f"Warning: 'Average Jitter' missing in {path}")
        else:
            Jitter.append(jit)

if PRR:
    plt.plot(PRR)
    plt.title("PRR")
    plt.show()
else:
    print("No PRR values found.")

if Throughput:
    plt.plot(Throughput)
    plt.title("Throughput")
    plt.show()
else:
    print("No Throughput values found.")

if Jitter:
    plt.plot(Jitter)
    plt.title("Jitter")
    plt.show()
else:
    print("No Jitter values found.")
```
